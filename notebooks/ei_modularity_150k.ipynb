# E/I modularity on the matched 150k set

Same analysis as section 6 of `lesion_exploration.ipynb`, on four budgets trained the same way
(`dale_all_woutfree_fanin_4pt_5seed_150k`: 150k steps, W_out trained, fanin init, batched), 5 seeds each:

| point | rate budget | wiring budget |
|---|---|---|
| control | none | none |
| A3 (rate only) | 0.0011 | free |
| B0 (wiring only) | free | 0.0054 |
| C0 (both) | 0.0115 | 0.0044 |

The three kinds of modularity, and the views that tell them apart:

1. **E modules + shared inhibition**: I neurons broad, cluster poorly; no I↔E pairing in wiring.
2. **Mixed E+I modules**: each I group matches an E group's tasks and is wired to it (wiring vs similarity ρ > 0).
3. **Inhibition between modules**: I groups wired to E groups of *other* tasks (ρ < 0).

Lesions (an I group silenced alone vs random I neurons) say whether the I part of a module is needed.

```python
import contextlib, io
import numpy as np
import pandas as pd
import torch
import matplotlib.pyplot as plt
from sklearn.cluster import KMeans
from sklearn.metrics import silhouette_score

from cmc.paths import MOO_ZOOM_RUNS_DIR
from cmc.task import generate_trials, rule_name
from cmc.train_cog import (batch_accuracy, eval_config_from, load_checkpoint, response_mask, ring_prefs,
                           trial_to_tensors)

RUN = MOO_ZOOM_RUNS_DIR / "dale_all_rateray_woutfrozen_150k"
NETS = {
    "control": RUN / "control" / "seed_01.pt",
    "rate-limited (A0)": RUN / "A0" / "seed_00.pt",
    "wiring-limited (B0)": MOO_ZOOM_RUNS_DIR / "dale_all_wiringray_woutfrozen_150k" / "B0" / "seed_00.pt",
}
EVAL_SEEDS = (10000, 10001, 10002)   # fixed eval trials (first 3 of the front's 10 eval seeds)
BATCH = 64
N_RANDOM = 5                         # random lesions (same E and I counts) per cluster
SIL_REF = 0.2
```

```python
def simulate(model, x, keep):
    # keep: (N,) tensor, 1 = normal, 0 = silenced
    x = model._normalize_input(x)
    B, _, T = x.shape
    h = torch.zeros(B, model.n_rnn)
    W = model.effective_w_rec()
    out = []
    for t in range(T):
        gate = x[:, :, t] @ model.w_in + h @ W + model.bias
        h = (1 - model.alpha) * h + model.alpha * model._cell_act(gate)
        h = h * keep
        out.append(torch.sigmoid(h[:, :model.n_e] @ model.w_out + model.b_out))
    return torch.stack(out, dim=1)


def cluster_units(tv, N):
    total = tv.sum(0)
    active = total > 1e-2 * total.max()
    norm = tv[:, active] / tv[:, active].max(0, keepdims=True)
    X = norm.T
    sil, labs = {}, {}
    for k in range(2, 21):
        labs[k] = KMeans(n_clusters=k, n_init=10, random_state=0).fit_predict(X)
        sil[k] = silhouette_score(X, labs[k])
    local = [k for k in range(3, 20) if sil[k] > sil[k - 1] and sil[k] > sil[k + 1]]
    K = max(local, key=sil.get)
    lab = labs[K]
    # order clusters by preferred task so the plots read top-left to bottom-right
    pref = {c: int(np.argmax(norm[:, lab == c].mean(axis=1))) for c in range(K)}
    order = sorted(range(K), key=lambda c: (pref[c], -norm[pref[c], lab == c].mean()))
    cluster = np.full(N, -1)
    cluster[np.flatnonzero(active)] = [order.index(c) for c in lab]
    return active, cluster, K, sil, local


def study(name, ckpt):
    with contextlib.redirect_stdout(io.StringIO()):
        model, config, ck = load_checkpoint(ckpt)
    model.eval()
    tasks = list(ck["active_tasks"])
    N, N_E = model.n_rnn, model.n_e
    trials = {}
    for rule in tasks:
        for s in EVAL_SEEDS:
            cfg = eval_config_from(config, s, easy_task=True)
            tr = generate_trials(rule, cfg, BATCH, noise_on=False)
            x, _, _, y_loc = trial_to_tensors(tr, "cpu")
            trials[rule, s] = (x, y_loc, response_mask(tr, "cpu"), ring_prefs(cfg, "cpu", torch.float32))

    @torch.no_grad()
    def task_acc(keep):
        acc = {}
        for rule in tasks:
            a = [batch_accuracy(simulate(model, x, keep), y_loc, prefs, resp_mask=rmask)["acc"]
                 for x, y_loc, rmask, prefs in (trials[rule, s] for s in EVAL_SEEDS)]
            acc[rule] = float(np.mean(a))
        return pd.Series(acc)

    base = task_acc(torch.ones(N))
    tv = np.asarray(ck["task_variance"], dtype=float)                 # (tasks, units)
    active, cluster, K, sil, local = cluster_units(tv, N)
    is_E = np.arange(N) < N_E
    act_E, act_I = np.flatnonzero(active & is_E), np.flatnonzero(active & ~is_E)
    rng = np.random.default_rng(0)
    drop, rand_drop, rows = {}, {}, []
    for c in range(K):
        members = np.flatnonzero(cluster == c)
        nE, nI = int(is_E[members].sum()), int((~is_E[members]).sum())
        keep = torch.ones(N); keep[members] = 0
        drop[c + 1] = base - task_acc(keep)
        r = []
        for _ in range(N_RANDOM):
            keep = torch.ones(N)
            keep[np.r_[rng.choice(act_E, nE, replace=False), rng.choice(act_I, nI, replace=False)]] = 0
            r.append(base - task_acc(keep))
        rand_drop[c + 1] = pd.concat(r, axis=1).mean(axis=1)
        rows.append(dict(cluster=c + 1, units=len(members), E=nE, I=nI))
    D = pd.DataFrame(drop).T.rename(columns=rule_name)
    RD = pd.DataFrame(rand_drop).T.rename(columns=rule_name)
    info = pd.DataFrame(rows).set_index("cluster")
    summary = info.assign(**{
        "mean drop": D.mean(axis=1), "random mean drop": RD.mean(axis=1),
        "excess over random": D.mean(axis=1) - RD.mean(axis=1),
        "max drop": D.max(axis=1), "worst task": D.idxmax(axis=1),
        "max specific drop": (D - RD).max(axis=1), "on task": (D - RD).idxmax(axis=1),
        "specificity": D.max(axis=1) - D.mean(axis=1)})
    print(f"{name}: {ckpt.relative_to(MOO_ZOOM_RUNS_DIR)} | intact min {base.min():.3f}, mean {base.mean():.3f} | "
          f"{active.sum()} active units, k = {K}, silhouette {sil[K]:.3f} (local maxima {local})")
    return dict(name=name, tasks=tasks, N=N, N_E=N_E, tv=tv, base=base, active=active, cluster=cluster, K=K,
                sil=sil, local=local, D=D, RD=RD, info=info, summary=summary)


def plot_clusters(R):
    tv, cluster, K, N_E = R["tv"], R["cluster"], R["K"], R["N_E"]
    labels = [rule_name[t] for t in R["tasks"]]
    fig, (ax, ax2) = plt.subplots(1, 2, figsize=(15, 4.5), gridspec_kw=dict(width_ratios=[3.2, 1]))
    idx = np.flatnonzero(R["active"])
    idx = idx[np.lexsort((idx >= N_E, cluster[idx]))]
    ax.imshow((tv / np.maximum(tv.max(0, keepdims=True), 1e-30))[:, idx], cmap="hot", aspect="auto",
              interpolation="nearest", vmin=0, vmax=1)
    ax.set_yticks(range(len(labels))); ax.set_yticklabels(labels, fontsize=7); ax.set_xticks([])
    cols = plt.cm.tab20(np.linspace(0, 1, 20))
    for c in range(K):
        pos = np.flatnonzero(cluster[idx] == c)
        ax.plot([pos[0] - .5, pos[-1] + .5], [len(labels) + .4] * 2, lw=5, color=cols[c % 20], clip_on=False,
                solid_capstyle="butt")
        ax.text(pos.mean(), len(labels) + 1.4, str(c + 1), ha="center", va="top", fontsize=7)
    ax.set_ylim(len(labels) - .5, -.5)
    ax.set_title(f"{R['name']}: task variance, units grouped by cluster (k = {K})", fontsize=9)
    sil, local = R["sil"], R["local"]
    ax2.plot(list(sil), list(sil.values()), "o-", ms=3)
    ax2.scatter(local, [sil[k] for k in local], s=60, facecolors="none", edgecolors="orange", label="local maxima")
    ax2.axvline(K, color="orange", ls="--", lw=1); ax2.axhline(SIL_REF, color="k", ls="--", lw=1)
    ax2.set_xlabel("k"); ax2.set_ylabel("silhouette"); ax2.legend(fontsize=7)
    plt.tight_layout(); plt.show()


def lesion_panel(ax, M, info, cmap, lim, title, annotate):
    # tasks on y, clusters on x; each cluster's column is as wide as its number of units (like the TV heatmap)
    edges = np.r_[0, np.cumsum(info.units.values)]
    tasks = list(M.columns)
    im = ax.pcolormesh(edges, np.arange(len(tasks) + 1) - 0.5, M.values.T, cmap=cmap, vmin=lim[0], vmax=lim[1])
    for x in edges[1:-1]:
        ax.axvline(x, color="k", lw=0.6)
    centres = (edges[:-1] + edges[1:]) / 2
    if annotate:
        for j, (c, w) in enumerate(zip(centres, info.units.values)):
            for i, v in enumerate(M.values[j]):
                if abs(v) >= 0.1 and w >= 6:
                    ax.text(c, i, f"{v:.2f}", ha="center", va="center", fontsize=6 if w >= 12 else 5)
    ax.set_yticks(range(len(tasks))); ax.set_yticklabels(tasks, fontsize=7)
    ax.set_ylim(len(tasks) - 0.5, -0.5)
    ax.set_xticks(centres)
    ax.set_xticklabels([f"{c}" for c in info.index], fontsize=7)
    ax.set_xlabel("cluster (column width = number of units)", fontsize=8)
    ax.set_title(title, fontsize=10)
    return im


def plot_lesions(R):
    D, RD, info = R["D"], R["RD"], R["info"]
    vmax = max(0.5, D.values.max())
    # small: the two raw drops
    fig, axes = plt.subplots(1, 2, figsize=(13, 4.6), sharey=True)
    for ax, M, title in [(axes[0], D, "accuracy drop when the cluster is silenced"),
                         (axes[1], RD, f"random lesion, same E and I counts (mean of {N_RANDOM})")]:
        im = lesion_panel(ax, M, info, "Reds", (0, vmax), title, annotate=False)
    fig.colorbar(im, ax=axes, fraction=0.02, pad=0.01, label="accuracy drop")
    fig.suptitle(R["name"], fontsize=11)
    plt.show()
    # big: specific drop
    fig, ax = plt.subplots(figsize=(15, 6.5))
    im = lesion_panel(ax, D - RD, info, "RdBu_r", (-0.5, 0.5),
                      f"{R['name']}: specific drop = cluster silenced − random (same E/I counts); "
                      f"red = the cluster matters for this task more than any group of its size", annotate=True)
    ax.set_xticklabels([f"{c}\n{r.units} u\n{r.E}E {r.I}I" for c, r in info.iterrows()], fontsize=7)
    fig.colorbar(im, ax=ax, fraction=0.02, pad=0.01, label="specific accuracy drop")
    plt.tight_layout(); plt.show()
```

```python
from scipy.stats import spearmanr

N_SHUFFLE = 500
PRESENT = 1e-2          # a synapse is present if its E/I-normalised |w| is above this (the conn_frac threshold)


def cluster_pop(norm, k_max):
    # norm: (tasks, units) of one population, each unit divided by its own max. Local-max rule, k > 2.
    X = norm.T
    k_max = min(k_max, len(X) - 1)
    sil, labs = {}, {}
    for k in range(2, k_max + 1):
        labs[k] = KMeans(n_clusters=k, n_init=10, random_state=0).fit_predict(X)
        sil[k] = silhouette_score(X, labs[k])
    local = [k for k in range(3, k_max) if sil[k] > sil[k - 1] and sil[k] > sil[k + 1]]
    K = max(local, key=sil.get) if local else max(sil, key=sil.get)
    lab = labs[K]
    pref = {c: int(np.argmax(norm[:, lab == c].mean(axis=1))) for c in range(K)}
    order = sorted(range(K), key=lambda c: (pref[c], -norm[pref[c], lab == c].mean()))
    return np.array([order.index(c) for c in lab]), K, sil, local


def ei_study(name, ckpt):
    with contextlib.redirect_stdout(io.StringIO()):
        model, config, ck = load_checkpoint(ckpt)
    model.eval()
    tasks = list(ck["active_tasks"])
    N, N_E = model.n_rnn, model.n_e
    trials = {}
    for rule in tasks:
        for s in EVAL_SEEDS:
            cfg = eval_config_from(config, s, easy_task=True)
            tr = generate_trials(rule, cfg, BATCH, noise_on=False)
            x, _, _, y_loc = trial_to_tensors(tr, "cpu")
            trials[rule, s] = (x, y_loc, response_mask(tr, "cpu"), ring_prefs(cfg, "cpu", torch.float32))

    @torch.no_grad()
    def task_acc(keep):
        return pd.Series({rule: float(np.mean([
            batch_accuracy(simulate(model, x, keep), y_loc, prefs, resp_mask=rmask)["acc"]
            for x, y_loc, rmask, prefs in (trials[rule, s] for s in EVAL_SEEDS)])) for rule in tasks})

    base = task_acc(torch.ones(N))
    tv = np.asarray(ck["task_variance"], dtype=float)
    total = tv.sum(0)
    active = total > 1e-2 * total.max()
    normfull = tv / np.maximum(tv.max(0, keepdims=True), 1e-30)
    out = dict(name=name, tasks=tasks, base=base, N=N, N_E=N_E, normfull=normfull)
    for pop, sel, k_max in (("E", np.arange(N) < N_E, 20), ("I", np.arange(N) >= N_E, 10)):
        units = np.flatnonzero(active & sel)
        lab, K, sil, local = cluster_pop(normfull[:, units], k_max)
        prof = np.stack([normfull[:, units[lab == c]].mean(axis=1) for c in range(K)])     # (K, tasks)
        breadth = (normfull[:, units] > 0.5).sum(0)                                       # tasks per neuron
        out[pop] = dict(units=units, lab=lab, K=K, sil=sil, local=local, prof=prof, breadth=breadth,
                        sizes=np.bincount(lab, minlength=K))

    # --- wiring between I groups and E groups (mean |w|, pruned = 0), z-scored against shuffled group labels
    W = model.effective_w_rec().detach().numpy()
    scale = model.ei_row_scale().detach().numpy()[:, None]
    A = np.where(np.abs(W) / scale > PRESENT, np.abs(W), 0.0)
    E, I = out["E"], out["I"]

    def blocks(labE, labI):
        GE = np.eye(E["K"])[labE]          # (nE, KE)
        GI = np.eye(I["K"])[labI]
        A_IE = A[np.ix_(I["units"], E["units"])]
        A_EI = A[np.ix_(E["units"], I["units"])]
        nE, nI = GE.sum(0), GI.sum(0)
        return (GI.T @ A_IE @ GE) / np.outer(nI, nE), (GE.T @ A_EI @ GI) / np.outer(nE, nI)

    IE, EI = blocks(E["lab"], I["lab"])
    rng = np.random.default_rng(0)
    nulls = [blocks(rng.permutation(E["lab"]), rng.permutation(I["lab"])) for _ in range(N_SHUFFLE)]
    nIE, nEI = np.array([n[0] for n in nulls]), np.array([n[1] for n in nulls])
    out["IE_z"] = (IE - nIE.mean(0)) / np.maximum(nIE.std(0), 1e-12)          # (KI, KE)
    out["EI_z"] = (EI - nEI.mean(0)) / np.maximum(nEI.std(0), 1e-12)          # (KE, KI)
    out["sim"] = np.corrcoef(np.vstack([I["prof"], E["prof"]]))[:I["K"], I["K"]:]   # (KI, KE) task-profile r

    # --- lesions: each E group alone vs random E neurons of the same number; same for I groups
    for pop in ("E", "I"):
        P = out[pop]
        drop, rdrop = {}, {}
        for c in range(P["K"]):
            members = P["units"][P["lab"] == c]
            keep = torch.ones(N); keep[members] = 0
            drop[c + 1] = base - task_acc(keep)
            r = []
            for _ in range(N_RANDOM):
                keep = torch.ones(N); keep[rng.choice(P["units"], len(members), replace=False)] = 0
                r.append(base - task_acc(keep))
            rdrop[c + 1] = pd.concat(r, axis=1).mean(axis=1)
        P["D"] = pd.DataFrame(drop).T.rename(columns=rule_name)
        P["RD"] = pd.DataFrame(rdrop).T.rename(columns=rule_name)
        P["spec"] = P["D"] - P["RD"]
        # does silencing a group hurt its OWN tasks? r(specific drop over tasks, the group's task profile)
        P["own_r"] = np.array([spearmanr(P["spec"].values[c], P["prof"][c])[0] for c in range(P["K"])])
    print(f"{name}: E {len(E['units'])} active, k = {E['K']} (sil {E['sil'][E['K']]:.3f}); "
          f"I {len(I['units'])} active, k = {I['K']} (sil {I['sil'][I['K']]:.3f})")
    return out


def plot_ei(R):
    E, I, labels = R["E"], R["I"], [rule_name[t] for t in R["tasks"]]
    # (a) task variance of E and I separately, units grouped by their own clusters; silhouette curves
    fig = plt.figure(figsize=(20, 4.8))
    gsp = fig.add_gridspec(1, 4, width_ratios=[len(E["units"]), max(len(I["units"]), 40), 45, 45])
    cols = plt.cm.tab20(np.linspace(0, 1, 20))
    for j, (pop, P) in enumerate((("E", E), ("I", I))):
        ax = fig.add_subplot(gsp[j])
        o = np.argsort(P["lab"], kind="stable")
        ax.imshow(R["normfull"][:, P["units"][o]], cmap="hot", aspect="auto", interpolation="nearest", vmin=0, vmax=1)
        ax.set_yticks(range(len(labels))); ax.set_yticklabels(labels if j == 0 else [], fontsize=7); ax.set_xticks([])
        labs = P["lab"][o]
        for c in range(P["K"]):
            pos = np.flatnonzero(labs == c)
            ax.plot([pos[0] - .5, pos[-1] + .5], [len(labels) + .4] * 2, lw=5, color=cols[c % 20], clip_on=False,
                    solid_capstyle="butt")
            ax.text(pos.mean(), len(labels) + 1.4, str(c + 1), ha="center", va="top", fontsize=7)
        ax.set_ylim(len(labels) - .5, -.5)
        ax.set_title(f"{pop} units ({len(P['units'])} active): k = {P['K']}, silhouette {P['sil'][P['K']]:.2f}\n"
                     f"tasks per neuron (norm. TV > 0.5): {P['breadth'].mean():.1f}", fontsize=9)
    for j, (pop, P) in enumerate((("E", E), ("I", I))):
        ax = fig.add_subplot(gsp[2 + j])
        ax.plot(list(P["sil"]), list(P["sil"].values()), "o-", ms=3)
        ax.axvline(P["K"], color="orange", ls="--", lw=1); ax.axhline(SIL_REF, color="k", ls="--", lw=1)
        ax.set_title(f"{pop}: silhouette vs k", fontsize=9); ax.set_xlabel("k")
    fig.suptitle(R["name"] + ": E and I clustered separately", fontsize=11)
    plt.tight_layout(); plt.show()

    # (b) task-profile similarity and wiring, I groups (rows) x E groups (cols)
    fig, axes = plt.subplots(1, 3, figsize=(18, 0.35 * I["K"] + 3.2))
    xl = [f"E{c + 1}\n{n}" for c, n in enumerate(E["sizes"])]
    yl = [f"I{c + 1} ({n})" for c, n in enumerate(I["sizes"])]
    for ax, M, title, lim in [(axes[0], R["sim"], "task-profile similarity (r) of I group and E group", 1),
                              (axes[1], R["IE_z"], "I → E wiring (z vs shuffled groups)", 4),
                              (axes[2], R["EI_z"].T, "E → I wiring (z vs shuffled groups)", 4)]:
        im = ax.imshow(M, cmap="RdBu_r", vmin=-lim, vmax=lim, aspect="auto")
        for (i, j), v in np.ndenumerate(M):
            if abs(v) >= (0.5 if lim == 1 else 2):
                ax.text(j, i, f"{v:.1f}", ha="center", va="center", fontsize=6)
        ax.set_xticks(range(E["K"])); ax.set_xticklabels(xl, fontsize=6)
        ax.set_yticks(range(I["K"])); ax.set_yticklabels(yl, fontsize=7)
        ax.set_title(title, fontsize=9)
        fig.colorbar(im, ax=ax, fraction=0.03)
    fig.suptitle(R["name"] + ": do I groups pair with E groups that do the same tasks, and are they wired to them?",
                 fontsize=11)
    plt.tight_layout(); plt.show()

    # (c) specific drop (group silenced - random neurons of the same population), tasks on y, width = size
    fig, axes = plt.subplots(1, 2, figsize=(18, 6), gridspec_kw=dict(width_ratios=[2.2, 1]))
    for ax, (pop, P) in zip(axes, (("E", E), ("I", I))):
        info = pd.DataFrame({"units": P["sizes"]}, index=range(1, P["K"] + 1))
        im = lesion_panel(ax, P["spec"], info, "RdBu_r", (-0.5, 0.5),
                          f"{pop} groups silenced alone − random {pop} neurons (same number)", annotate=True)
        ax.set_xticklabels([f"{pop}{c}\n{n}" for c, n in info.units.items()], fontsize=7)
        if pop == "I":
            ax.set_yticklabels([])
    fig.colorbar(im, ax=axes, fraction=0.02, pad=0.01, label="specific accuracy drop")
    fig.suptitle(R["name"], fontsize=11)
    plt.show()


def ei_summary(R):
    E, I = R["E"], R["I"]
    pairs_IE = spearmanr(R["sim"].ravel(), R["IE_z"].ravel())[0]
    pairs_EI = spearmanr(R["sim"].ravel(), R["EI_z"].T.ravel())[0]
    return {
        "E: tasks per neuron": E["breadth"].mean(), "I: tasks per neuron": I["breadth"].mean(),
        "E silhouette": E["sil"][E["K"]], "I silhouette": I["sil"][I["K"]],
        "best E match per I group (mean r)": R["sim"].max(axis=1).mean(),
        "I→E wiring vs similarity (ρ)": pairs_IE, "E→I wiring vs similarity (ρ)": pairs_EI,
        "I lesion hits own tasks (mean ρ)": np.nanmean(I["own_r"]),
        "E lesion hits own tasks (mean ρ)": np.nanmean(E["own_r"]),
        "I groups, mean |specific drop|": I["spec"].abs().values.mean(),
        "E groups, mean |specific drop|": E["spec"].abs().values.mean(),
    }
```

## 1. Run all 20 networks

About 2-3 minutes per network.

```python
RUN150 = MOO_ZOOM_RUNS_DIR / "dale_all_woutfree_fanin_4pt_5seed_150k"
POINTS = ["control", "A3", "B0", "C0"]
LABEL = {"control": "control", "A3": "A3 rate only", "B0": "B0 wiring only", "C0": "C0 both"}
runs150 = pd.read_csv(RUN150 / "runs.csv")
ALL = {}
for p in POINTS:
    for s in range(5):
        ALL[p, s] = ei_study(f"{LABEL[p]} seed {s}", RUN150 / p / f"seed_{s:02d}.pt")
```

## 4. Reading it (mean ± sd over 5 seeds)

| | control | A3 rate only | B0 wiring only | C0 both |
|---|---|---|---|---|
| E tasks per neuron | 4.7 | **2.9** | 5.2 | 3.5 |
| I tasks per neuron | 4.8 | 4.1 | 4.8 | 4.0 |
| E silhouette | 0.19 | **0.28** | 0.23 | 0.24 |
| I silhouette | 0.24 | **0.31** | 0.24 | 0.26 |
| I→E wiring vs task similarity (ρ) | **−0.44 ± 0.16** | 0.01 ± 0.04 | +0.19 ± 0.19 | +0.18 ± 0.09 |
| E→I wiring vs task similarity (ρ) | +0.04 ± 0.15 | 0.01 ± 0.12 | +0.20 ± 0.17 | **+0.46 ± 0.13** |
| I lesion hits own tasks (ρ) | 0.30 | 0.16 | **0.46 ± 0.07** | 0.11 |

* **control → inhibition between modules (type 3 direction).** In all 5 seeds, I groups are wired *less* to the E
  groups that share their tasks and more to the others (ρ −0.31 to −0.71). E.g. seed 1: I4 (17 units) shares the
  tasks of E3-E6 (r 0.6-0.9) but sends them fewer/weaker synapses than chance (z −2 to −3.6). I neurons are not
  broader than E (4.8 vs 4.7 tasks), so it is not "shared inhibition" either.
* **A3 (rate only) → E+I modules in tuning, not in wiring.** The sharpest tuning of all (E neurons in 2.9 tasks,
  silhouettes 0.28 / 0.31), but wiring is unrelated to the tasks (ρ ≈ 0 ± 0.04 in every seed): the recurrent
  weights stay dense and close to their random start. Lesions do the least damage of all.
* **B0 (wiring only) → weak E+I modules, and the I part matters.** Wiring leans towards matching groups (ρ ≈ +0.2,
  seed-dependent), E neurons are broad (5.2 tasks), and silencing an I group hits its *own* tasks most clearly of
  any point (ρ 0.46 ± 0.07, every seed 0.39-0.55).
* **C0 (both) → E+I modules in tuning and in wiring.** E→I wiring follows the tasks in every seed (+0.32 to +0.63),
  I→E too (+0.10 to +0.33). The I part of a module is not individually needed (ρ 0.11).

**So:** the inhibition changes role with the budget. Without a budget it is *competitive* (I groups hold down E
groups of other tasks). A rate budget alone sharpens tuning without touching the wiring. A wiring budget (alone or
with rate) wires I neurons *into* their own E groups: E+I modules, most clearly when both budgets act (C0).


# 5. Methods

Everything above is computed by the code cells of this notebook; this section describes what they do, in order.
Nothing here was re-run after the results were produced.

## 5.1 Networks

* Run `runs/moo_zoom/dale_all_woutfree_fanin_4pt_5seed_150k`, produced by
  `python -m cmc.moo_zoom --task-battery all --steps 150000 --n-seeds 5 --w-out-init fanin --batched --batch-pop 20
  --points "control=none; A3=0.001107,1; B0=10,0.005363; C0=0.01148,0.004387"`.
* Dale RNN, 256 units (204 E, 52 I), 20 Yang (2019) tasks, 150 000 training steps, batch 32, W_in, W_rec and W_out
  all trained (W_out initialised as randn / √204, "fanin"). All 20 networks were trained in the same batched run,
  so they share one trial stream; they differ only in budget and seed.
* Budgets (rate = learned-multiplier constraint on the L2 firing-rate cost; wiring = L1-ball projection of W_rec
  after every step):

  | point | rate budget | wiring budget | meaning |
  |---|---|---|---|
  | control | none | none | unconstrained |
  | A3 | 0.0011 | 1 (never binds) | rate only |
  | B0 | 10 (never binds) | 0.0054 | wiring only |
  | C0 | 0.0115 | 0.0044 | both |

* min_task_acc (worst of the 20 tasks) comes from the run's `runs.csv` (10 eval seeds, noise-free). Every network
  is above 0.6.

## 5.2 Task variance (as in Yang et al. 2019)

* Stored in each checkpoint (`task_variance`, computed after training by `cmc.runner.activity_summary`): for each
  task, 64 noise-free trials on each of the 10 eval seeds; for each unit, the variance of its activity **across
  trials** at each time step after fixation onset, averaged over time, then over eval seeds. Result: a
  (20 tasks × 256 units) matrix TV.
* **Active units**: total task variance (sum over tasks) > 1 % of the network's largest total. Inactive units are
  left out of all clustering, matching and lesions.
* **Normalisation**: each unit's 20 values are divided by that unit's own maximum, so every unit's profile lies in
  [0, 1] and the most-varied task is 1.
* **Tasks per neuron** ("breadth"): the number of tasks with normalised TV > 0.5, averaged over the active units of a
  population.

## 5.3 Clustering E and I separately

* The active E units and the active I units are clustered **separately**, each with k-means on the normalised
  20-dimensional profiles (scikit-learn, `n_init = 10`, `random_state = 0`).
* k from 2 to 20 for E and from 2 to 10 for I (never more than the number of units − 1); silhouette score for each k.
* **Choice of k**: the largest silhouette among the *local maxima* with k > 2 (silhouette at k higher than at k − 1
  and at k + 1). If no such local maximum exists, the plain maximum is used.
* Clusters are re-numbered by their preferred task (the task with the highest mean normalised TV in the cluster) so
  plots read in task order.
* **Task profile of a group**: the mean normalised TV of its units, a 20-vector.
* **E silhouette / I silhouette** in the tables: the silhouette at the chosen k.

## 5.4 Do I groups pair with E groups? (task similarity)

* For every I group and every E group: Pearson r between their two task profiles (20 tasks). Matrix `sim`
  (I groups × E groups), left panel of figure (b).
* **Best E match per I group**: for each I group the highest r over E groups, averaged over I groups. High values
  mean every I group has an E group doing the same tasks.

## 5.5 Wiring between groups

* W = the effective recurrent matrix `model.effective_w_rec()`, W[pre, post] (rows = presynaptic). A synapse
  counts as **present** if |W| / (E/I row scale) > 0.01 (the `conn_frac` threshold; the row scale is 1 for E rows
  and n_E / n_I for I rows, so E and I are judged on the same footing). A = |W| for present synapses, 0 otherwise
  (so a pruned synapse counts as 0).
* **Block strength**: for I group a and E group b, the mean of A over all (pre ∈ a, post ∈ b) pairs: the I → E
  block. Likewise E → I with E rows and I columns. This combines how many synapses exist and how strong they are.
* **Null**: the E group labels are shuffled among the active E units and the I labels among the active I units
  (group sizes kept), 500 times, and the block strengths recomputed.
  **z = (observed − mean of shuffles) / sd of shuffles**. Middle (I → E) and right (E → I) panels of figure (b).
* **Wiring vs similarity (ρ)**: Spearman correlation, over all (I group, E group) pairs of one network, between
  the task similarity r and the wiring z. Positive: groups doing the same tasks are wired together more than
  chance; negative: groups doing *different* tasks are wired together more; about 0: wiring ignores the tasks.
  Computed separately for I → E and E → I.

## 5.6 Lesions

* **Silencing**: the network is run with its own weights, but after every time step the activity of the silenced
  units is set to 0, for the whole trial. A silenced unit therefore sends nothing to other units or to the readout.
  (Same update as `DaleRNN.simulate`: h ← (1 − α) h + α relu(x W_in + h W_rec + b), then h ← h · keep.)
* **Accuracy**: each of the 20 tasks scored with the project's `batch_accuracy` (circular decoding, fixation
  release), on 64 noise-free trials from each of 3 eval seeds (10000-10002). The same trials are reused for every
  lesion of a network, so differences come only from the lesion.
* **Drop**: accuracy of the intact network − accuracy with the group silenced, per task.
* **Random control**: the same number of units drawn at random from the *same population* (active E units for an E
  group, active I units for an I group), 5 draws, drops averaged.
* **Specific drop** = drop (group) − drop (random), per task. Figure (c), tasks on y, groups on x, each column as
  wide as the group. Red: the group matters for that task more than any group of its size from its population.
* **Lesion hits own tasks (ρ)**: for each group, the Spearman correlation over the 20 tasks between its specific
  drop and its own task profile; averaged over the groups of a population. Positive: silencing a group hurts the
  tasks it is tuned to.
* **Mean |specific drop|**: the average absolute specific drop over groups and tasks (how much group identity
  matters at all).

## 5.7 How the numbers map to the three kinds of modularity

| | type 1: E modules + shared I | type 2: mixed E+I modules | type 3: inhibition between modules |
|---|---|---|---|
| I tasks per neuron vs E | I clearly broader | similar | similar |
| best E match per I group | low / irrelevant | high | high |
| wiring vs similarity (ρ) | ≈ 0 | > 0 | < 0 (I → E) |
| I lesion hits own tasks | weak | > 0 if the I part is needed | other tasks |

Reading of the results (section 4) with this table:

* **control**: I → E ρ = −0.44 ± 0.16, negative in every seed (−0.31 to −0.71); I not broader than E (4.8 vs 4.7
  tasks) → type 3 direction.
* **A3**: sharpest tuning (E 2.9 tasks, silhouette 0.28), best E match 0.79, but ρ ≈ 0 for both directions in every
  seed → E+I modules in tuning only.
* **B0**: ρ ≈ +0.2 (seed-dependent), I lesion own-task ρ 0.46 ± 0.07 (every seed 0.39-0.55) → weak type 2 with a
  necessary I part.
* **C0**: E → I ρ +0.46 ± 0.13 (every seed +0.32 to +0.63), I → E +0.18 ± 0.09 (every seed > 0) → type 2 in
  tuning and wiring; I part not individually needed (ρ 0.11).

## 5.8 Caveats

* 5 seeds per point; each network's wiring ρ rests on (number of I groups × number of E groups) pairs, about 15-170.
  Differences between points are only claimed where the seeds barely overlap (control vs the rest for I → E; C0 vs
  the rest for E → I; B0 vs the rest for I lesion own-task).
* k is chosen per network and per population, and varies across seeds; all numbers above are per-group averages and
  do not depend on matching groups across networks.
* In the dense networks (control, A3) almost every synapse is present, so their wiring z-scores reflect weight
  *strength* only; in B0 and C0 (about 1-5 % of synapses present) they also reflect which synapses exist.
* Lesions set activity to exactly 0 (a strong lesion); the random control uses the same population and number, but
  not the same activity level.
* All analysis is on the final (150k) networks; no network was retrained or re-evaluated for this notebook beyond
  the lesion simulations.
